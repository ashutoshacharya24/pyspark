# Increamental Loading with Auto Loader

#### Suppose we have source --> sink. In source we have one file(file_1) so it moved that file_1 to sink, then next time file_2 & file_3 came up in source, now it will only move the new 2 files to sink.
#### it creates checkpoints (one is schema refered by source, 2nd one is it will source folder checkpoint at sink and it store all the metadata in dbrocks folder)

**Streaming DF**

```python
from pyspark.sql.functions import *
```

```python
df = spark.readStream.format("cloudFiles")\
            .option('cloudFiles.format', 'parquet')\
                .option('cloudFiles.schemaLocation', 'abfss://mycontainer@storagemoderndbashu.dfs.core.windows.net/autosink/check')
                    .load('abfss://mycontainer@storagemoderndbashu.dfs.core.windows.net/autosource')
```

```python
df.writeStream.format("parquet")\
        .option('checkpointLocation', 'abfss://mycontainer@storagemoderndbashu.dfs.core.windows.net/autosink/check')\
            .trigger(processingTime = '10 seconds')\
                .start('abfss://mycontainer@storagemoderndbashu.dfs.core.windows.net/autosink/data')
```
