# What is Schema?
### Schema is just  the logical structure that we are picking inside data model.
## 1. star schema(When we have one fact table and dimensiona tables are sorrounding it, it becomes star schema. There is no heirarchy between the dimensions)
## 2. Snowflake schema(When we have one fact table and dimensiona tables are sorrounding it, it becomes star schema. There is heirarchy between the dimensions)

# Types of Fact Tables
### 1. Granular/Transaction Fact Table(1 Transaction = 1 Row)
### 2. Periodic Fact Table(1 row = monthly Transaction)
### 3. Accumulating Fact Table(so many date cols, 1 row = Process)

# Types of Dimension tables
### 1. Conformed Dimensions(if the dimension table connected to multiple fact table)
### 2. Role Playing Dimensions(When we connected our fact table with dimension table with mulitple conditions)
### 3. Junk Dimension(when we have one or two data in dimension table then it is called junk dimension)
### 4. Degenerate Dimension(when we don't have anything to map with the col)

# Slowly Changing Dimensions
